```python
while True:
    try:
        number = int(input('양의 정수 입력'))
        if number <= 0:
            raise ValueError
    except ValueError:
        print('양의 정수를 입력하세요.')
    else:
        sum = 0
        for i in range(1, number+1):
            sum += i
        print(f'>> 1부터 {number}까지의 합 : {sum}')
        break
```

### 학습 내용 되돌아보기 (2)

```python
try:
    m = int(input('월(month) 입력 > '))
    if not (1 <= m <= 12):
        raise Exception('1부터 12까지의 정수를 입력하세요.')
    
except Exception as e:
    print('예외 발생: {}'.format(e))

else:
    print('%d월: 잘 입력했습니다.' % m)
```

### 학습 내용 되돌아보기 (3)

```python
def divide(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print('0으로는 나눌 수 없습니다.')
    else:
        print(f' 결과: {result}')

divide(3.2, 2)
divide(5.4, 0)
```

### 학습 내용 되돌아보기 (4)

```python
uyear = {1:'freshman', 2:'sophomore', 3:'junior', 4:'senior'}
year = int(input('대학교 몇 학년이지요?'))

try:
    if year not in uyear:
        raise Exception('1~4 정수를 입력하세요.')
except Exception as e:
    print(f'예외 발생 이름: {type(e)}')
    print(f'예외 발생 이유: {e}')
else:
    print(f'{year}학년: {uyear[year]}')
finally:
    print(f'예외 처리가 잘되는군요!')
```
