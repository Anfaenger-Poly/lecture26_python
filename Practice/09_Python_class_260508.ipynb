### 파이썬 클래스

```python
class Circle:
    pi = 3.14
    def __init__(self, radius):
        self.radius = radius # 왼쪽은 멤버변수, 오른쪽은 매개변수
    def area(self):
        return self.radius ** 2 * Circle.pi # 클래스 이름으로 접근
        
c = Circle(2.5)
print(c.area())
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
    
rec = Rectangle(2.1, 3.4)
print(rec.area())
```

```python
class Account:
    def __init__(self):
        self.balance = 0
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount): # 잔액이 부족한것도 체크해야함
        self.balance -= amount
    
a1 = Account()
a1.deposit(100)
print(a1.balance)
```

```python
class Account:
    def __init__(self):
        self.__balance = 0
    def deposit(self, amount):
        self.__balance += amount
    def withdraw(self, amount): # 잔액이 부족한것도 체크해야함
        self.__balance -= amount
    def get_balance(self): # get_balance()를 통해 private 멤버에 접근
        return self.__balance
    
a2 = Account()
a2.deposit(50)
print(a2.get_balance()) # 에러 발생, private 멤버는 외부에서 접근 불가
```

```python
# 매직 매서드 연습
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f'{self.name}({self.age}살)'
    def __eq__(self, other):
        if self.name == other.name and self.age == other.age:
            return True
        return False # else 생략 가능

p1 = Person('홍길동', 15)
print(p1)

p2 = Person('홍길동', 15)
print(p2)

if p1 == p2:
    print('같은 사람입니다')
else:
    print('다른 사람입니다.') # 객체를 따로 생성했기 때문에 다른사람으로 인식함
```

### 학습 내용 되돌아보기 (2)

```python
class Integer:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value
    
    def __sub__(self, other):
        return self.value - other.value
    
a = Integer(5)
b = Integer(3)

print(a + b)
print(a - b)
```

### 학습 내용 되돌아보기 (4)

```python
class Student:
    def __init__(self, name, dept, mid, final):
        self.name = name
        self.dept = dept
        self.mid = mid
        self.final = final
    
    def __str__(self):      # 점수는 숫자 -> 문자열로 str()처리
        return '학과: ' + self.dept + ', 이름: ' + self.name + \
            ', 중간점수: ' + str(self.mid) + ', 기말점수: ' + str(self.final)
    
    def grade(self):
        avg = (self.mid + self.final) / 2
        if avg >= 90:
            return 'A'
        elif avg >= 80:
            return 'B'
        elif avg >= 70:
            return 'C'
        elif avg >= 60:
            return 'D'
        else:
            return 'F'
        
s = Student('김경철', '기계학과', 89, 90)
print(s)
print('학점: ', s.grade())
    
    
```

### 학습 내용 되돌아보기 (5)

```python
class Car:
    def __init__(self, company, year, color):
        self.company = company
        self.year = year
        self.color = color
    
    def __str__(self):                         # 숫자라서 str()로 처리
        return '자동차 회사: ' + self.company + ', 년식: ' + str(self.year) + ', 색상: ' + self.color
    
    def __eq__(self, other):
        return self.company == other.company and self.year == other.year and self.color == other.color
    
mycar = Car('현대', 2020, '검정')
yourcar = Car('기아', 2021, '백색')
print(mycar)
print(yourcar)
print(mycar == yourcar)
```
