### 파이썬 클래스

```python
class Circle:
    pi = 3.14
    def __init__(self, radius):
        self.radius = radius # 왼쪽은 멤버변수, 오른쪽은 매개변수
    def area(self):
        return self.radius ** 2 * Circle.pi # 클래스 이름으로 접근
        
c = Circle(2.5)
print(c.area())
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
    
rec = Rectangle(2.1, 3.4)
print(rec.area())
```

```python
class Account:
    def __init__(self):
        self.balance = 0
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount): # 잔액이 부족한것도 체크해야함
        self.balance -= amount
    
a1 = Account()
a1.deposit(100)
print(a1.balance)
```

```python
class Account:
    def __init__(self):
        self.__balance = 0
    def deposit(self, amount):
        self.__balance += amount
    def withdraw(self, amount): # 잔액이 부족한것도 체크해야함
        self.__balance -= amount
    def get_balance(self): # get_balance()를 통해 private 멤버에 접근
        return self.__balance
    
a2 = Account()
a2.deposit(50)
print(a2.get_balance()) # 에러 발생, private 멤버는 외부에서 접근 불가
```
