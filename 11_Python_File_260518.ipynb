```python
filename = './sample.txt'

with open(filename, encoding='utf-8') as f_read:
    content = f_read.read() # f_read 벗어나면 자동으로 파일이 닫힘
print(content)
# with 구문을 벗어나면 자동으로 파일이 닫히므로, f_read.close()를 따로 호출할 필요가 없음
```

```python
with open(filename) as f_read:
    lines = f_read.readlines() # 파일의 모든 줄을 읽어서 리스트로 반환

for line in lines:
    print(line.strip()) # 각 줄의 양쪽 공백을 제거하여 출력
```

```python
with open(filename) as f_read:
    while True:
        line = f_read.readline() # 파일에서 한 줄씩 읽어서 반환
        if line: print(line.strip())
        else: break
```

```python
fruits = ['사과', '토마토', '바나나']
outfile = 'fruits.txt'
with open(outfile, mode='w', encoding='utf-8') as f_write:
    for fruit in fruits:
        # for fruit in fruits:
        #     f_write.write(fruit + '\n') , 각 과일 이름 뒤에 줄바꿈 문자 추가
        f_write.write('\n'.join(fruits)) # 리스트의 요소들을 줄바꿈 문자로 연결하여 한 번에 쓰기)

with open(outfile, mode='a', encoding='utf-8') as f_write:
    f_write.write('\n'.join(fruits))
```

```python
with open(filename, mode='rb') as f_read:
    content = f_read.read() # 바이너리(16진수) 모드로 파일을 읽어서 bytes 객체로 반환 -> 한글로 변환 하려면 content.decode('utf-8') 사용
print(content)
print(content.decode('utf-8')) # bytes 객체를 utf-8 인코딩으로 디코딩하여 문자열로 변환하여 출력
```

### 학습 되돌아 보기 (2)

```python
import math
from random import random

def getarea(r):
    return round(math.pi * r ** 2, 2)

r = round(random() * 10, 2)

print(f'원의 반지름: {r}')
print(f'원주율 pi: {round(math.pi, 4)}')
print(f'반지름 {r}인 원의 면적은 {getarea(r)}')
```

### 학습 되돌아 보기 (3)

```python
import math
import statistics

def main():
    # factorial 출력
    for n in [1, 6, 11, 16]:
        print(f'{n}! = {math.factorial(n)}')
    print()

    st = [80, 99, 77, 65, 92, 74, 82]
    print(st)
    print(f'중앙 값: {statistics.median(st):.2f}')
    print(f'평균: {statistics.mean(st):.2f}')
    print(f'분산: {statistics.variance(st):.2f}')
    print(f'표준편차: {statistics.stdev(st):.2f}')

if __name__ == '__main__':
    main()
```
